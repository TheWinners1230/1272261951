```python
a = int(input("A ="))
b = int(input("B ="))
c = int(input("C ="))

def right_angled_triangle_checker(a,b,c):
    if (a**2 == b**2 + c**2):
        print("Right Angled Triangle!")
    elif (b**2 == a**2 + c**2):
        print("Right Angled Triangle!")
    elif(c**2 == b**2 + a**2):
        print("Right Angled Triangle!")
    else:
        print("Not Right Angled Triangle!")
right_angled_triangle_checker(a,b,c)
```
