```python
print("Current List:-")
list = ["Toyota","Volkswagen","Ford","Hyundai","Honda"]
print(list)
print()
print("Number of Items in list are:-" , len(list))
print()
print("Adding Koenigsegg to the list:-")
list.append("Koenigsegg")
print(list)
print()
list.insert(2, "BMW")
print("Inserting BMW in between Volkswagen and Ford:-")
print(list)
print()
print("Deleting Hyundai fromt list:-")
list.pop(4)
print(list)
print()
print("Slicing the list from Volkswagen till Honda:-")
print(list[1:5])
print()
print("Deleting entire list:-")
del list
print(list)
```

```python
tuple = ("Toyota","Volkswagen","Ford","Hyundai","Honda")
print(tuple)
print()
print("Number of Items in list are:-" , len(tuple))
print()
print("Adding Koenigsegg to the list:-")
y = ("Koenigsegg",)
tuple += y
print(tuple)
print()
print("Deleting tuple:-")
del tuple
print(tuple) #this will raise an error because the tuple no longer exists
```

```python
dict = {"brand": "Lamborghini", "model": "Murcielago", "year": 2001}
print(dict)
print()
print("Adding color green:-")
dict["color"] = "green" #Can also use update(.)
print(dict)
print()
print("Removing model:-")
dict.pop("model")
print(dict)
print()
print("Deleting entire Dictionary:-")
del dict
print(dict)
```
