```python
a = int(input("Enter the first number (A): "))
b = int(input("Enter the second number (B): "))
c = int(input("Enter the third number (C): "))

if a == b == c:
    print(f"All three numbers are equal, which is {a}.")
elif a == b and a > c:
    print(f"A and B are equal and are the largest: {a}")
elif a == c and a > b:
    print(f"A and C are equal and are the largest: {a}")
elif b == c and b > a:
    print(f"B and C are equal and are the largest: {b}")
elif a > b and a > c:
    print(f"A is the largest number: {a}")
elif b > a and b > c:
    print(f"B is the largest number: {b}")
else:
    print(f"C is the largest distinct number: {c}")
```
